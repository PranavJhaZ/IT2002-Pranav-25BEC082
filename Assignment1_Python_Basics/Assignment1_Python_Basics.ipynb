### Q3. Strings
Write a Python program to accept a string and: Count the number of characters, Count vowels and consonants, Reverse the string, Check whether it is a palindrome.

```python
user_input = input("Enter any string: ")
print("Length:", len(user_input))

v_set = set("aeiouAEIOU")
vowels = 0
consonants = 0

for character in user_input:
    if character.isalpha():
        if character in v_set:
            vowels += 1
        else:
            consonants += 1

print("Vowel count:", vowels)
print("Consonant count:", consonants)

rev_str = user_input[::-1]
print("Reversed string:", rev_str)

if user_input.lower() == rev_str.lower():
    print("Result: String is a palindrome")
else:
    print("Result: String is not a palindrome")
```

### Q4. Sets and Dictionaries
Create two sets of integers and demonstrate union, intersection, difference, and symmetric difference. Also create a dictionary containing the names and marks of five students and display the student with the highest marks.

```python
A = {100, 200, 300, 400, 500}
B = {400, 500, 600, 700, 800}

print("Set A:", A)
print("Set B:", B)
print("Union:", A.union(B))
print("Intersection:", A.intersection(B))
print("Difference (A - B):", A.difference(B))
print("Symmetric Difference:", A.symmetric_difference(B))

grades = {"Pranav": 92, "Rohan": 81, "Simran": 95, "Aditya": 78, "Ananya": 89}
highest_student = max(grades, key=grades.get)
print(f"Highest marks obtained by: {highest_student} ({grades[highest_student]} marks)")
```

### Q5. GPA Calculator
Develop a Python program that accepts the grade/grade point and credit of different courses and calculates the student's SGPA/GPA using: GPA = Σ(Grade Point × Credit) / ΣCredits

```python
num_subs = int(input("Enter number of courses: "))
weighted_total = 0.0
credit_sum = 0.0

for k in range(num_subs):
    gp = float(input(f"Enter grade point for subject {k + 1}: "))
    c = float(input(f"Enter credit for subject {k + 1}: "))
    weighted_total += gp * c
    credit_sum += c

calculated_gpa = weighted_total / credit_sum if credit_sum != 0 else 0.0
print(f"Calculated SGPA: {calculated_gpa:.2f}")
```

### Q6. Functions and Iteration
Write separate Python functions to: Calculate the factorial of a number, Generate the Fibonacci series up to n terms, Check whether a number is prime.

```python
def compute_factorial(number):
    if number < 0:
        return None
    res = 1
    for step in range(2, number + 1):
        res *= step
    return res

def make_fibonacci(count):
    if count <= 0:
        return []
    elif count == 1:
        return [0]
    fib_list = [0, 1]
    while len(fib_list) < count:
        fib_list.append(fib_list[-1] + fib_list[-2])
    return fib_list

def check_prime(number):
    if number <= 1:
        return False
    divisor = 2
    while divisor * divisor <= number:
        if number % divisor == 0:
            return False
        divisor += 1
    return True

print("Factorial of 6:", compute_factorial(6))
print("Fibonacci (8 terms):", make_fibonacci(8))
print("Is 29 prime?", check_prime(29))
print("Is 35 prime?", check_prime(35))
```

### Q9. List Comprehension
Using list comprehension, generate: Squares of numbers from 1 to 20, Even numbers from 1 to 50, Numbers divisible by both 3 and 5 from 1 to 100.

```python
sq_numbers = [k ** 2 for k in range(1, 21)]
even_numbers = [k for k in range(2, 51, 2)]
div_3_5 = [k for k in range(1, 101) if k % 15 == 0]

print("Squares:", sq_numbers)
print("Evens:", even_numbers)
print("Divisible by 3 and 5:", div_3_5)
```

### Q10. Mini Programming Problem
Develop a menu-driven Python program for maintaining a student record using a dictionary. The program should provide options to: Add a student, Update marks, Search for a student, Delete a student, Display all student records.

```python
student_db = {}

while True:
    print("\n1. Add Student\n2. Update Marks\n3. Search Student\n4. Delete Student\n5. Show All\n6. Exit")
    opt = input("Enter choice (1-6): ").strip()

    if opt == "1":
        name = input("Enter name: ").strip()
        score = float(input("Enter marks: "))
        student_db[name] = score
        print("Student added.")
    elif opt == "2":
        name = input("Enter name to update: ").strip()
        if name in student_db:
            student_db[name] = float(input("Enter new marks: "))
            print("Marks updated.")
        else:
            print("Student not found.")
    elif opt == "3":
        name = input("Enter name to search: ").strip()
        if name in student_db:
            print(f"{name} -> {student_db[name]}")
        else:
            print("Student not found.")
    elif opt == "4":
        name = input("Enter name to delete: ").strip()
        if name in student_db:
            del student_db[name]
            print(f"Record for {name} removed.")
        else:
            print("Student not found.")
    elif opt == "5":
        if not student_db:
            print("Database is empty.")
        else:
            for s_name, s_marks in student_db.items():
                print(f"{s_name}: {s_marks}")
    elif opt == "6":
        print("Exiting...")
        break
    else:
        print("Invalid choice, please select between 1 and 6.")
```
